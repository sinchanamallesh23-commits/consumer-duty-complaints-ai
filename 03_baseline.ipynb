```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score, classification_report

train = pd.read_csv("data/clean/train.csv")
test = pd.read_csv("data/clean/test_1000.csv")

vec = TfidfVectorizer(max_features=50000, ngram_range=(1, 2), min_df=3, stop_words="english")
X_train = vec.fit_transform(train["narrative"])
X_test = vec.transform(test["narrative"])

clf = LogisticRegression(max_iter=2000)
clf.fit(X_train, train["issue"])
test["baseline_pred"] = clf.predict(X_test)
```

```python
majority = test["issue"].value_counts(normalize=True).iloc[0]
print(f"Guess the most common issue every time: {majority*100:.1f}% accuracy")
print(f"Baseline accuracy:  {accuracy_score(test['issue'], test['baseline_pred'])*100:.1f}%")
print(f"Baseline macro F1:  {f1_score(test['issue'], test['baseline_pred'], average='macro')*100:.1f}%")
print()
print(classification_report(test["issue"], test["baseline_pred"], digits=2))

import os
os.makedirs("outputs", exist_ok=True)
test[["complaint_id", "issue", "baseline_pred"]].to_csv("outputs/baseline_predictions.csv", index=False)
```
